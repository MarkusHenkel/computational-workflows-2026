1. What is nf-core?

nf-core is a community-curated set of bioinformatics pipelines built with NextFlow that follows guidelines and best practices. 

It also offers a template for creating your own workflow and offers modules of certain bioinformatics tools that can be imported into your own workflow which simplifies the building process.

2. How many pipelines are there currently in nf-core?

Per https://nf-co.re/pipelines/:
- Released pipelines: 101
- Pipelines under development: 43
- Archived pipelines: 13


3. Are there any non-bioinformatic pipelines in nf-core?

Yes, there is 'datasync' which is a pipeline for copying files and directories between storage locations with checksums for file validation.

4. Let's go back a couple of steps. What is a pipeline and what do we use it for?

A pipeline is a series of processes where within the pipeline the output of individual processes is fed into following processes.

This way instead of having to manually take each output and inputting it in the next process, we can automate this. This makes executing a single command for completing a whole workflow possible.

5. Why do you think nf-core adheres to strict guidelines?

Strict guidelines enforce workflows whose code can be understood more easily, that function in predictable ways and have consistent structure.

Following these guidelines thus makes nf-core a collection of standardized pipelines that non-domain experts can also use and that community members can maintain more easily. 



6. What are the main features of nf-core pipelines?

nf-core ...
- ... pipelines run consistently across laptops, HPC clusters, and cloud environments (interoperability)
- ... provides templates, tools and guidelines for building standardized Nextflow pipelines and thus pipelines have consistent structure and adhere to best practices
- ... is maintained by a global community so that ideally pipelines stay useful and are not deprecated
- ... offers version control so that previous versions of a pipeline can also be used (reproducibility)
(per https://nf-co.re/docs/get_started/nf-core)

### Differential abundance analysis
#### Description
Differential abundance analysis is a statistical method for testing if the expression levels of genes between two or more groups differ significantly more than expected from biological and technical variability. In our case there are two groups according to the resulting plots,  mCherry and hND6.  


Going by the names of the two groups, mCherry is a fluorescent protein and used in fluoresence microscopy as an intracellular probe. Group two (hND6) apparently refers to a human mitochondrial gene. The comparison here could thus be treatment vs. control where mCherry is the control and hND6, the 'treatment', a gene that was delivered to a subject e.g. by viral transduction. A differential abundance analysis may then be able to show the effects of the gene.
