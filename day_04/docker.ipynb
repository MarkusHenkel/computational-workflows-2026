### What is a container?

A container is an architecture that encapsulates an application and all of its dependencies. 

### Why do we use containers?

Since the container contains the application and all of its dependencies, it runs consistently and independently of the host sytem and in any environment. This is especially useful in cases where an application is developed on a developer's laptop, tested in a testing environment and and deployed to a production server.

It ensures that the correct versions of tools within the applications are used and thus it ensures reproducibility.

### What is a docker image?

A docker image is an executable software package that includes everything needed to run an application. This includes: code, libraries, settings and a runtime. Executing the docker image with `docker run` creates a running docker container and starts the application.

1. create a container holding fastqc using seqera containers (https://seqera.io/containers/)
2. use the container to generate a fastqc html of the example fastq file

````python
# pull the container
!docker run community.wave.seqera.io/library/fastqc:0.12.1--9971ea336a9eddae

# Output

# Unable to find image 'community.wave.seqera.io/library/fastqc:0.12.1--9971ea336a9eddae' locally
# 0.12.1--9971ea336a9eddae: Pulling from library/fastqc
# ed8f0ce1546f: Pull complete 
# Digest: sha256:f97f7658a191bebe2f2a007cab5b815d380f89a948d6eb1481b046ab7618bebc
# Status: Downloaded newer image for community.wave.seqera.io/library/fastqc:0.12.1--9971ea336a9eddae
````

````python
# run the container and save the results to a new "fastqc_results" directory

# with the -it flag an interactive shell is created
!docker run --rm -it community.wave.seqera.io/library/fastqc:0.12.1--9971ea336a9eddae

# alternatively: remove -it and just put the fastqc command at the end, e.g.
!docker run \
    -v "/home/markus/MasterVault/Semester 2/biomedical_data/computational-workflows-2026/day_02/SRA_runs/fastq/SRX19144486_SRR23195516_1.fastq.gz:/input.fastq.gz" \
    -v "$(pwd)/fastqc_results:/results" \
    --rm community.wave.seqera.io/library/fastqc:0.12.1--9971ea336a9eddae \
    fastqc -o results input.fastq.gz

# -v "$(pwd)/fastqc_results:/results": mount a host directory inside the container so that host output dir is visible inside the container
# same goes for the input file
````

### Now that you know how to use a docker container, which approach was easier and which approach will be easier in the future?

The docker approach. 

This is because manually downloading files, unzipping them, making them executable is tedious in comparison with just getting an image with one command and then running it using a second command.

However, correctly mounting the correct input file and output dir was hard initially whereas I already knew how to download an application manually. Since I now know how mounting works, I expect the docker approach to be easier in the future.

### What would you say, which approach is more reproducible?

Since in both approaches both the image and manual download contain the same FastQC version, both results should be identical. 
Both approaches can reproduce the result, however if someone else now wanted to reproduce these results and downloaded FastQC manually there's a risk of them using the wrong version and potentially getting different results. If I instead provide this person with the image that I used then they would certainly use the correct FastQC version and thus the docker container approach is more reproducible.
